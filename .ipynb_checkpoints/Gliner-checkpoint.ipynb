```python
import pandas as pd
from gliner import GLiNER
from tqdm import tqdm
from datetime import datetime
from transformers import AutoTokenizer
```

```python
# Path to your dataset file
file_path = 'sample-1M.jsonl'

# Read only first 100k lines
df = pd.read_json(file_path, lines=True)

print(f"Successfully loaded {len(df)} articles.\n")

df[['title', 'content']].head()

```

```python

# Load GLiNER model
model = GLiNER.from_pretrained("urchade/gliner_base")
# tokenizer = AutoTokenizer.from_pretrained("urchade/gliner_base")  # for token counting
```

```python
print(df.columns)
```

```python
# ----- Start Time -----
start_time = datetime.now()
print("NER started at:", start_time)

# Load GLiNER model
model = GLiNER.from_pretrained("urchade/gliner_base")

# NER labels
labels = [
    # Core OntoNotes
    "person", "norp", "facility", "organization", "gpe", "location",
    "product", "event", "work_of_art", "law", "language",
    "date", "time", "percent", "money", "quantity", "ordinal", "cardinal",
    # Common extensions
    "religion", "political_party", "nationality", "ethnic_group",
    "title", "award", "disease", "chemical", "weapon",
    "vehicle", "currency", "brand"
]

def batch_ner(texts, batch_size=16):
    """Run NER directly on each text without chunking."""
    results = []
    texts = texts.fillna("").astype(str).tolist()

    for i in tqdm(range(0, len(texts), batch_size)):
        batch = texts[i:i+batch_size]
        preds = model.run(batch, labels, batch_size=len(batch))
        results.extend(preds)

    return results

title = df["title"].fillna("")
# content = df["content"].fillna("")


# Run NER
entities = batch_ner(title, batch_size=16)

# Create new dataframe with just id + entities
entities_df = pd.DataFrame({
    "id": df["id"],
    "entities": entities
})

# ----- End Time -----
end_time = datetime.now()
print("NER finished at:", end_time)
print("Total time taken:", end_time - start_time)

# Show first rows
entities_df.head()
```

```python
output_path = "ner_result_1M.json"

entities_df[["id", "entities"]].to_json(
    output_path,
    orient="records",
    indent=2,
    force_ascii=False
)

# print(f"NER results saved to {output_path}")

# entities_df.head()
```
